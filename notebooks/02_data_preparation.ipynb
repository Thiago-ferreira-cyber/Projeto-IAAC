# Data Preparation

Fase de limpeza e preparação dos dados, usando as funções de `data_preparation.py`.

```python
from pathlib import Path
import sys

SRC_PATH = Path("..").resolve() / "src"
OUTPUT_DIR = Path(".").resolve()
sys.path.insert(0, str(SRC_PATH))

import data_preparation as dp

```

## 1. Carregar o dataset

```python
DATA_PATH = Path("..").resolve() / "dataset" / "raw" / "MalwareMemoryDump.csv"
df = dp.load_data(str(DATA_PATH))
```

## 2. Valores nulos e duplicados

```python
dp.check_missing_duplicates(df)
```

## 3. Limpeza (tipos de dados + remoção de identificadores)

```python
df_clean = dp.clean_data(df, id_column="Raw_Type")
df_clean.shape
df_clean.dtypes
```

## 4. Separar features (X) e target (y)

```python
TARGET = "Label"
X, y = dp.split_features_target(df_clean, TARGET)
```

## 5. Dividir em treino e teste

```python
X_train, X_val, X_test, y_train, y_val, y_test = dp.split_train_validation_test(
    X,
    y,
    validation_size=0.15,
    test_size=0.15,
    random_state=42,
)
```
